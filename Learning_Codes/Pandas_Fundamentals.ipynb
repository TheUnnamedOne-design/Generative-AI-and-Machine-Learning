```python
import pandas as pd
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
dataset = fetch_ucirepo(id=350) 
  
# data (as pandas dataframes) 

X = dataset.data.features 
y = dataset.data.targets 
  
# Combine into one DataFrame
df = pd.concat([X, y], axis=1)

print(df.head())
```

```python
# -- Handling missing data --------------------------------
df.dropna()  # drop rows with any NaN

df.dropna(subset=["X6"])  # drop rows where income is NaN

df.fillna(0)  # replace NaN with 0

df["X7"] = df["X7"].fillna(df["X7"].median())




```

```python
# -- Dropping columns -------------------------------------
df.drop(columns=["X2", "X3"], inplace=True)
```

```python
# -- Renaming ---------------------------------------------
df.rename(columns={"X1": "LIMIT_BAL"}, inplace=True)
df.rename(columns={"LIMIT_BAL": "X1"}, inplace=True)

df.describe()
```
